**Task:**

Write three functions that generate numerical arrays using NumPy.

Each function takes three parameters:

* **start** – the starting value of the array

* **stop** – the ending value of the array (not inclusive)

* **steps** – the increment between values in the array (the step size used by `np.arange`)

Use `np.arange(start, stop, steps)` to generate the base array of values. Then apply the requested NumPy math function to the entire array.

**Functions to Implement:**

1. `generate_sine_wave(start, stop, steps)`

* Generate a NumPy array of values from `start` to `stop` using `np.arange`.
* Compute the sine of each value using `np.sin`.
* Return the array of sine values.
2. `generate_logs(start, stop, steps)`
* Generate a NumPy array of values from `start` to `stop` using `np.arange`.
* Compute the natural logarithm of each value using `np.log`.
* Return the array of log values.
3. `generate_rounded(start, stop, steps)`
* Generate a NumPy array of values from `start` to `stop` using `np.arange`.
* Round each value using `np.round`.
* Return the rounded array.



````python
# Your Solution
def generate_sine_wave(start, stop, steps):
    "Compute the sine of each value"
    x = np.arange(start, stop, steps)
    return np.sin(x)

def generate_logs(start, stop, steps):
    "Compute the logarithm of each value"
    y = np.arange(start, stop, steps)
    return np.log(y)

def generate_rounded(start, stop, steps):
    "Compute the rounded array of each value"
    z = np.arange(start, stop, steps)
    return np.round(z)
````

**Problem 2 (review)**

**Concept:** *reading CSV file, create an np.array of numbers*

**Task:**

Write a function `read_csv_to_dict(filename)` that reads a CSV file and stores the data in a dictionary.

Your function should:
1. Open the CSV file.
2. Read the first row to obtain column headers.
3. Store each column of data as a NumPy array.
4. Use the column names as keys in the dictionary.
5. Return the dictionary.

Each key should correspond to one column in the CSV file.

Example structure:

```
{
   "open": np.array([...]),
   "close": np.array([...]),
   "volume": np.array([...])
}
```


````python
#Your Solution
def read_csv_to_dict(filename):
    """Reads a CSV file and stores each column in a dictionary as a NumPy array."""
    with open(filename, "r") as file:
        rows = [line.strip().split(",") for line in file if line.strip()]

    header = rows[0]
    data = {}

    for i, name in enumerate(header):
        column = [row[i] for row in rows[1:]]
        data[name] = np.array(column)

    return data
````

Steps:

1. Generate `x` values between **0 and 1**.

$$y = sqrt(1 − x²)$$

3. Reflect the points across all four quadrants so that the full circle is drawn.

````python
#Your Solution
"Create a scatter plot that displays the points of a unit circle."
x = np.linspace(0, 1, 101) 
y = np.sqrt(1 - (x**2))

#get all four quadrants
x_all = np.concatenate([x, -x, -x, x])
y_all = np.concatenate([y, y, -y, -y])

# Create the plot
plt.figure(figsize=(5, 5))
plt.scatter(x_all, y_all, color='orange', marker='o', s=8)

# 3. Add labels and title
plt.title('Unit Circle Scatter Plot')
plt.xlabel('x')
plt.ylabel('y')
plt.axis('equal') 
````

**Problem 10**

**Concepts:** *reading CSV file, line graph*

**Task:**

Write a function:

`create_line_graph(filename, column_name)`

Your function should:
1. Read the CSV file using `read_csv_to_dict`.
2. Verify the specified column exists.
3. Convert the column values to floats.
4. Plot the values as a **line graph**.

Your graph should include:
* a title
* x-axis labeled **Day Number**
* y-axis labeled with the column name

````python
#Your Solution
def create_line_graph(filename, column_name):
    """Reads a CSV file and creates a line graph of one column."""
    data = read_csv_to_dict(filename)

    # Verify the requested column exists
    if column_name not in data:
        raise ValueError(f"Column '{column_name}' not found in {filename}")

    # Convert the column values to floats
    values = data[column_name].astype(float)
    days = np.arange(len(values))

    # Create line graph
    plt.figure(figsize=(10, 5))
    plt.plot(days, values, color='green', linewidth=2.5, label=column_name)
    plt.margins(y=0.1)

    # Labels and a title
    plt.title(f'Line Graph of {filename}', fontsize=14)
    plt.xlabel('Number of Days', fontsize=12)
    plt.ylabel(column_name, fontsize=12)
    plt.grid(True)
    plt.legend(loc='upper right', fontsize=8)
    plt.tight_layout()
````

**Problem 11**

**Concepts:** *reading CSV file, histogram*

**Task:**

Write a function:

`create_histogram(filename, column_name)`

Your function should:
1. Read the CSV file using `read_csv_to_dict`.
2. Convert the selected column to numeric values.
3. Create a histogram of the data.

Requirements:
* Use **at least 10 bins**.
* Add a title showing the column name.

The histogram should display the distribution of values in the selected column.

````python
#Your Solution
def create_histogram(filename, column_name):
    """Reads a CSV file and creates a bar histogram of one column."""
    # Read the CSV file into a dictionary of columns
    data = read_csv_to_dict(filename)

    # Convert the column values to floats
    values = data[column_name].astype(float)

    # Count the values in each of 10 bins
    counts, bin_edges = np.histogram(values, bins=10)
    bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2   # middle of each bin
    bin_width = bin_edges[1] - bin_edges[0]              # width of each bin

    # Draw one bar per bin, with a small gap between bars
    fig, ax = plt.subplots(figsize=(11, 6))
    bars = ax.bar(bin_centers, counts, width=bin_width * 0.9, color='orange', edgecolor='black', label=column_name)
    ax.bar_label(bars, fontsize=9)

    ax.set_xticks(bin_edges)
    ax.set_xticklabels([f'{edge:,.0f}' for edge in bin_edges], rotation=45)
    ax.set_xlim(bin_edges[0] - bin_width * 0.5, bin_edges[-1] + bin_width * 0.5)
    ax.set_ylim(0, counts.max() * 1.05)

    # Labels and a title
    ax.set_title(f'Histogram of {column_name} from {filename}', fontsize=14)
    ax.set_xlabel(column_name, fontsize=12)
    ax.set_ylabel('Frequency', fontsize=12)
    ax.grid(True, axis='y')
    ax.legend(loc='upper right', fontsize=9)
    plt.tight_layout()
````

**Problem 12**

**Concepts:** *reading CSV file, pie chart*

**Task:**

Write a function:

`create_pie_chart(filename, column_name)`

Your function should:
1. Read the CSV file using `read_csv_to_dict`.
2. Convert the selected column to numeric values.
3. Compute the **mean value** of the column.
4. Determine:
* the percentage of values **greater than the mean**
* the percentage **less than or equal to the mean**

Create a pie chart displaying these two proportions.

Include:
* labels
* percentages
* chart title

````python
#Your Solution
def create_pie_chart(filename, column_name):
    """Reads a CSV file and creates a pie chart of values above and at or below the mean."""
    # Read the CSV file into a dictionary of columns
    data = read_csv_to_dict(filename)

    # Convert the column values to floats
    values = data[column_name].astype(float)

    # Compute the mean and count values above and at or below it
    mean_value = np.mean(values)
    greater_count = np.sum(values > mean_value)
    less_equal_count = np.sum(values <= mean_value)

    sizes = [greater_count, less_equal_count]
    labels = ['Greater than mean', 'Less than or equal to mean']
    colors = ['orange', 'pink']
    explode = (0.1, 0)  # pull out the first slice slightly

    #pie chart
    fig, ax = plt.subplots(figsize=(6, 6))
    ax.pie(
        sizes,
        labels=labels,
        autopct='%1.1f%%',         # show percentages inside slices
        startangle=90,             # start at the top
        explode=explode,
        colors=colors,
        shadow=True,
        wedgeprops={'edgecolor': 'white', 'linewidth': 1}
    )

    ax.set_title(f'Relative Value for Mean {mean_value:.2f}', fontsize=14)
    plt.tight_layout()
````

````python
create_pie_chart("SBUX.csv", "close")
````
